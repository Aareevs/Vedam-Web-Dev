# Groupby in Pandas

Groupby in Pandas is a powerful tool that allows you to group data based on one or more columns and perform aggregate functions on those groups. 

```python
tip = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/tip.csv')
titanic = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/Titanic-Dataset.csv')
planets = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/planets.csv')
bookings = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/bookings.csv')
mpg = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/mpg.csv')
```

```python
tip.groupby("day")["total_bill"].size()
```

Aggregate functions are functions that take a list of values and return a single value. For example, the mean function takes a list of numbers and returns their average. In pandas, you can use aggregate functions to summarize data in a DataFrame.

```python
tip.groupby("day").agg(
    bill_count = ("total_bill", "size"),
    bill_mean = ("total_bill", "mean"),
    avg_tip = ("tip", "mean")
)
```

```python
tip.groupby("time")["total_bill"].size()    
```

```python
# Summary is a way of aggregating data based on a specific column, in this case, "day". The aggregation functions used here are:
# - "size": counts the number of rows in each group
# - "mean": calculates the average of the values in each group

summary = tip.groupby("day").agg(
    bill_count = ("total_bill", "size"),
    bill_mean = ("total_bill", "mean"),
    avg_tip = ("tip", "mean")
).round(2)

summary
```

```python
# To convert the summary DataFrame to a CSV file, we can use the `to_csv()` method.

summary.to_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/tip_summary.csv')
```
