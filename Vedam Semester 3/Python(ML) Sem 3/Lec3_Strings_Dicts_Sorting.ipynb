**Questions**

Q1) User enters their full name carelessly: " rahul KUMAR sharma ". Clean it and print it as "Rahul Kumar Sharma". Also Validate that name contains only alphabets and spaces, print 'invalid name' other wise.

Q2) Marks are given as one messy string: 45, 88, 72, 91, 60.

Convert it into a list of integers, then print the total and average marks.

Q3) Validate an email entered by uer. Rules:

-> clean the spaces and convert to lowercase

-> must contain only one **@**

-> must end with **.com**

-> For part before @ must be alphanumeric. Print **Valid email** or **invalid email**

Q4) For a given sentence:

-> print total number of words

-> print how many words start with a vowel

```python
# Q1 
name = " rahul KUMAR sharma "
name = name.strip().title()
if all(x.isalpha() or x.isspace() for x in name):
    print(name)
else:
    print("invalid name")
```

```python
# Q2 

marks = "45, 88, 72, 91, 60"
marks_list = [int(mark.strip()) for mark in marks.split(",")]
total_marks = sum(marks_list)
average_marks = total_marks / len(marks_list)
print("Total Marks:", total_marks)
print("Average Marks:", average_marks)
```

```python
# Q3 

email = input().strip()
if email.count("@") == 1 and email.endswith(".com"):
    local_part = email.split("@")[0]
    if local_part.isalnum():
        print("Valid email")
    else:
        print("Invalid email part before @ must be alphanumeric.")
else:
    print("Invalid email: Must contain one '@' and end with '.com'.")
```

## Dictionary
### setdefault(key, default) - get + insert if missing

```python
d = {"Aareev": 28, "Awaneesh": 3, "Shivansh":20}
d.setdefault("Aareev", 0)
```

### Views: keys(), values(), items()

```python
print(d.keys())
print(d.values())
print(d.items())
```

### update()

```python
d = {"Aareev": 28, "Awaneesh": 3, "Shivansh":20}
d["Shivansh"] = 21
print(d)
d.update({"Aareev": 1, "Awaneesh": 2})
print(d)
```

### pop(), popitem(), del and clear() - Deleting Methods

```python
d = {"Aareev": 28, "Awaneesh": 3, "Shivansh":20}
d.pop("Awaneesh")
print(d)
del d["Shivansh"]
print(d)
```

### fromkeys()

```python
marks = dict.fromkeys(["Math","English","Physics"], 0)
print(marks)
```

### Looping Patterns + membership (in)

```python
d = {"Aareev": 28, "Awaneesh": 3, "Shivansh":20}
for k, v in d.items():
    print(k,v)
```

### Nestted Dict Access

```python
college = {
    "CS": {"HOD": "Dr. Torune", "students": 200},
    "AI": {"HOD": "Dr. Dave", "students": 150},
}

for branch in college.keys():
    print(college[branch]["HOD"])
```
