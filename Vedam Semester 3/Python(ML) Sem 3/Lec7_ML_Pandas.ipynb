# Pandas

**Pandas** is a key Python library used for data cleaning, organization, and preparation before building machine learning models. 

FLOW: Pandas = NumPy + labels -- Series(1D + Index) -- DataFrame(2d table) -- creating from a dict -- numpy sits underneath

```python
import pandas as pd
import numpy as np

# pd.DataFrame() - from a dictionary

restaraunt_data = {
    'Table No': [12, 14, 15],
    'Bill Amount': [1250, 890, 2400],
    'Waiter_Name': ['Shivansh', 'Awaneesh', 'Prity']
}

df = pd.DataFrame(restaraunt_data)

print(df)
print(type(df))
print(df['Bill Amount'])
print(type(df['Bill Amount']))
```

```python
# pd.Series() - 1D Data with an index

total_bill = [4700, 8000, 1200, 500] 
s = pd.Series(total_bill, index=['Mon', 'Tue', 'Wed', 'Thur'])
print(s)
print(type(s))
```
