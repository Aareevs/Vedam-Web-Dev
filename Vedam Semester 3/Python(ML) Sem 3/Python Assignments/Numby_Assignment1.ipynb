### Q6

Given a = np.array([2, 4, 6]) and b = np.array([2.0, 4.0, 6.0]) - check both dtypes, then convert b
to int with astype and verify the dtypes now match.

```python
# Q6  

a = np.array([2, 4, 6])
b = np.array([2.0, 4.0, 6.0])
print("Before:")
print("a dtype:", a.dtype)
print("b dtype:", b.dtype)

b = b.astype(int)
print("\nAfter:")
print("a dtype:", a.dtype)
print("b dtype:", b.dtype)
```

### Q15

Given the marks array below - (a) extract all marks greater than 60; (b) extract marks between 50 and 80 using two conditions with & (remember the parentheses!).

marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

```python
# Q15

marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

# (a) Marks greater than 60
print(marks[marks > 60])

# (b) Marks between 50 and 80
print(marks[(marks >= 50) & (marks <= 80)])
```
