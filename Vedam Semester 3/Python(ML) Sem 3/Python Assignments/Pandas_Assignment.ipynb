```python
# Q1) Load the dataset into t. How many rows and how many columns does it have?

t = pd.read_csv('Titanic-Dataset.csv')
print("Rows:", t.shape[0])
print("Columns:", t.shape[1])
```

```python
# Q2) Print all the column names. Pick any three and write in a comment what you think each one holds.

print("Column Names:")
for col in t.columns:
    print(col) 

# Comments on three columns:
# 'Survived' - This likely holds binary data (0 or 1) indicating if the passenger survived the sinking.
# 'Sex' - This contains categorical data ('male' or 'female') describing the passenger's biological sex.
# 'Fare' - This is numerical data representing the price of the ticket the passenger paid for the voyage.
```

```python
# Q4) Run t.info(). Some columns have fewer values than the total number of rows. Write down which 
# columns those are, and how many values are missing from each. 

t.info()

```

```python
# Q5: Run t.describe(). Which columns are left out and why?

t.describe()

# Two columns left out are 'Name' and 'Cabin'.
# Reason: The .describe() method by default only calculates statistics for numerical columns (int and float). 
# 'Name' and 'Cabin' are string/object data types, so they are excluded from the statistical summary.
```

```python
# Q6: For the class column, answer using a different function for each.

# a) How many different values are there?
print(t['Pclass'].nunique()) 

# b) What are they?
print(t['Pclass'].unique())  

# c) How many passengers fall into each one?
print(t['Pclass'].value_counts()) 
```

```python
# Q7: Find how many passengers were male and how many were female.

print(t['Sex'].value_counts())
```

```python
# Q8) In describe(), look at the age column. Compare its mean with its 50% value. Which is larger, and
# what does that suggest about the passengers' ages?

# From the output of t.describe():
# Age Mean = 29.699118
# Age 50% (Median) = 28.000000

# The Mean (29.7) is larger than the Median (28.0).
# Suggestion: This indicates the Age distribution is "right-skewed" (positively skewed). 
# It means there are a small number of older passengers (outliers on the higher end) 
# pulling the average age higher than the middle value of the dataset.
```

```python
# Q9 (Challenge) The age column has missing values. Using only shape and info(), work out what
# percentage of age values are missing. Write the number in a comment.

total_rows = t.shape[0]
non_null_age = 714  # Found from .info() output
missing_age = total_rows - non_null_age

percentage_missing = (missing_age / total_rows) * 100
print(f"{percentage_missing:.2f}%") 
```
