# NumPy

## What is NumPy

> **Flow**: Lists VS Arrays (speed, memory) -> **ndarray** -> the idea of vectorization -> Pandas/Sklearn are built on top of NumPy -> import as **np**

```python
#np.array() - 1D and 2D

import numpy as np
x = np.array([2,3,4,5])
x1 = np.array([[2,3],[4,5],[8,9]])
print(x)
print(x.dtype)
print(type(x))
print(x1)
```

```python
# np.zeros()

print(np.zeros(20))
print(np.zeros(10,dtype='int32'))
print(np.zeros((3,4), dtype='int16'))
```

```python
# np.ones()

print(np.ones(10))
print(np.ones((2,3,4)))
```

```python
#  np.full()

print(np.full((4,3), 15))
```

```python
# np.arange(start, stop, step)

print(np.arange(10,5,-2))
print(np.arange(10))
print(np.arange(1,11).reshape(2,5))
```

```python
# np.linspace(start, stop, num)

print(np.linspace(1,10,5))
print(np.linspace(0,1,5))
```

```python
# np.eye(n) - returns a 2-D array with ones on the diagonal and zeros elsewhere

print(np.eye(4))
# print(np.eye(3,4,k=2))
```

## Array Attributes

```python
# .shape() - reshape the array

x = np.arange(20).reshape(4,5)
print(x)
print(x.shape)
```

```python
# .ndim() - get the number of dimensions 

print(x.ndim)
```

```python
# 1D Indexing - Positive and Negative Indexing

arr=np.arange(10)
print(arr)
print(arr[-3:-12:-2])
```

```python
# 1D Slicing - [start:stop:step], reverse
```

```python
# 2D Indexing - arr[row, col], full row, full column

arr = np.arange(20).reshape(5,4)
print(arr)
print(arr[2,2])
```

```python
# 2D Slicing - Sub Matrix (block)

print(arr)
print(arr[1:4, 1:3])  # Slicing rows 1 to 3 and columns 1 to 2
print(arr[1,1:]) # Slicing row 1 and columns 1 to end
```

```python
# astype() - convert data type of array

x = np.array([1, 4, 4.3, 4.9, 4.8])
print(x)
print(x.astype('int32'))
print(x)
```
