### NumPy Fundamentals

> Flow: array 

```python
# arithmetic - array with scalar, array with array

import numpy as np

arr1 = np.arange(10)
print(arr1)
arr1=arr1 + 5
print("Multiply",arr1*5)
print("Exponent",arr1**2)
print(arr1)
```

```python
# Sum Arithmetic operations on two arrays

x = np.arange(10)
y = np.arange(11,21)
print(x)
print(y)
print("Sum",x+y)
```

```python
# comparison operators - boolean arrays

x = np.arange(5)
y = np.arange(1,6)
print(x==y)
print(x>y)
print(y>x)
```

### np.random()

> Flow: seed (why reproductibility matters) -> rand -> randn -> randint (high excluded) -> choice -> shuffle

```python
# np.random.seed() - set the random seed; np.random.rand() - generate random numbers

np.random.seed(2)
np.random.rand(5)
np.random.randn(10)
```

```python
# np.random.randint(low,high,size) - generate random integers

np.random.randint(10,20,size=(5,3))
```

```python
# np.random.choice()

lst = [3,4,5,6,7,8,9]
np.random.choice(lst, 3, replace=False)
np.random.choice(10,5,replace=False)
```

```python
# np.random.shuffle()

arr = np.arange(10)
np.random.shuffle(arr)
print(arr)
```
