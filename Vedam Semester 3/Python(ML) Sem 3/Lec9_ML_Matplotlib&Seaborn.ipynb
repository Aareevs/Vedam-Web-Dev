````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/Titanic-Dataset.csv')
mpg = pd.read_csv('/Users/aareev/VS-Code/Vedam/Vedam Semester 3/Python(ML) Sem 3/Data/mpg.csv')       \

print(df.shape, mpg.shape)
````

---
## The Matplotlib Skeleton

Three lines around every plot:

```python
plt.figure(figsize=(7, 4))   # size
sns.something(...)           # the plot
plt.title('...')             # label it
plt.show()
```

````python
# a plain histogram of Fare, with a size and a title -> histplot 

plt.figure(figsize=(7, 3))
sns.histplot(data=df, x='Fare')
plt.title('Fare')
plt.show()
````

````python
# plt.grid() - to show grid lines

plt.figure(figsize=(7, 4))
sns.histplot(data=df, x='Fare', y='Age', bins=30)
plt.title('Fare')
plt.grid()
plt.show()
````

````python
# plt.savefig('name.png') - to save the figure as a png file
# plt.savefig('name.png') must come before plt.show() — after it, the figure is gone.

plt.savefig('Fare_histogram.png')
````

````python
# histplot() - to show histogram of Fare with a size and a title
plt.figure(figsize=(5, 3))
sns.histplot(data=df, x='Fare')
plt.title('Distribution of Fare')
plt.show()
````

````python
# countplot() - to show count of each category in Embarked with a size and a title

plt.figure(figsize=(5, 3))
sns.countplot(data=df, x='Survived')
plt.title('Count of Survived and Died')
plt.show()
````

````python
# Q1) Plot the distribution of mpg. Is it skewed? If this column had blanks, would you fill with the mean or the median?

plt.figure(figsize=(5, 3))
sns.histplot(data=mpg, x='mpg')
plt.show()
````

````python
# Q2) How many cars come from each `origin`? What would that imbalance do to a model trained on this data?

plt.figure(figsize=(5, 3))
sns.countplot(data=mpg, x='origin')
plt.title('Count of Cars by Origin')
plt.show()
````
