### Question 1 · The Strict Username Validator

Write a function that takes a username string as input and validates it. To be valid, the username must be at least 5 characters long after removing any leading/trailing spaces, and it must contain only letters and numbers (no special characters like @, _, or spaces inside). If it is valid, return the clean, strictly lowercase version of the username. If it is invalid, return the string "Invalid".

```python
# Q1 

def validate_username(username):
    username = username.strip()
    if len(username) < 5:
        return "Invalid"
    if not username.isalnum():
        return "Invalid"
    return username.lower()

username = input().strip()
print(validate_username(username))
```

### Question 2 · The Spy Code Reverser

You intercept a secret code where words are separated by hyphens (-). To decode it, you must reverse the order of the words, convert every letter to UPPERCASE, and replace the hyphens with spaces.

```python
# Q2 

secret_code = input().strip()
decoded_code = secret_code.split('-')[::-1]
decoded_code = ' '.join(decoded_code).upper()
print(decoded_code)
```

### Question 3 · The Messy Database Cleaner

You have received a raw string of names from a database exported as a semicolon-separated list. The data is messy, full of random spaces, and has inconsistent casing. Write a function to clean this string so that every name is properly **Title Cased**, extra spaces are removed, and the names are joined into a single string separated by a comma and a space.

```python
# Q3 

def clean_names(raw_string):
    names = raw_string.split(';')
    cleaned_names = [name.strip().title() for name in names]
    return ', '.join(cleaned_names)

raw_string = input()
print(clean_names(raw_string))
```

### Question 5 · The Palindrome Sentence Checker

Determine if a given sentence is a palindrome. To do this correctly using only string methods, remove all spaces and non-alphanumeric characters, convert everything to lowercase, and check whether it reads the same forwards and backwards

```python
# Q5

sentence = input().strip()
cleaned_sentence = ''.join(char.lower() for char in sentence if char.isalnum())
if cleaned_sentence == cleaned_sentence[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

### Question 6 · The Hashtag Extractor

Given an input string representing a social media post, extract all the hashtags. The extracted hashtags must be completely lowercase, and the # symbol should be removed. Store them in a list. Ignore any words that are not hashtags.

```python
# Q6 

string = input().strip()
hashtags = [word[1:].lower() for word in string.split() if word.startswith('#')]
print(hashtags)
```

### Question 7 · The File Extension Finder

You are given a single comma-separated string containing various filenames. Extract just the file extensions (without the dot), ensure there are no duplicates, and return them as a list of strings strictly in UPPERCASE.

```python
# Q7 

filenames = input().strip().split(',')
extensions = list(set(filename.split('.')[-1].upper() for filename in filenames))
print(extensions)
```

### Question 8 · The Domain Name Slicer

Given an email address string, extract just the company/domain name — the part immediately after the @
symbol and before the dot suffix like .com or .edu. Convert the extracted domain to Title Case. If the string does not contain an @, return "Invalid".

```python
# Q8 
email = input().strip()
if '@' in email:
    domain = email.split('@')[1].split('.')[0]
    print(domain.title())
else:
    print("Invalid")
```

### Question 9 · The Anagram Detective

Write a function to check if two given strings are anagrams of each other — they contain the exact same letters in the exact same frequencies, just rearranged. Ignore any spaces and treat uppercase and lowercase letters as the same.
> Note: Solve it using string methods and basic built-in functions (like sorted())

```python
# Q9 Write a function to check if two given strings are anagrams of each other — they contain the exact same letters in the exact same frequencies, just rearranged. Ignore any spaces and treat uppercase and lowercase letters as the same. > Note: Solve it using string methods and basic built-in functions (like sorted())


def are_anagrams(str1, str2):
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()

    return sorted(str1) == sorted(str2)

str1 = input().strip()
str2 = input().strip()
if are_anagrams(str1, str2):
    print("True")
else:
    print("False")
```

### Question 10 · The Messy Phone Number Formatter

You are building a CRM system and user-submitted phone numbers are a mess — they contain spaces,
parentheses, periods, and hyphens. Write a function that takes a raw phone number string and cleans it by removing all non-digit characters. Once cleaned, validate the number:

• If the cleaned string contains exactly 10 digits, format and return it as "(XXX) XXX-XXXX".

• If it contains anything other than 10 digits, return "Invalid Number"

> Note: Case 1 fails because after removing symbols it has 11 digits, due to the leading “1”

```python
# Q10 

def clean_number (raw_number):
    cleaned_number = ''.join(filter(str.isdigit, raw_number))
    if len(cleaned_number) == 10:
        return f"({cleaned_number[:3]}) {cleaned_number[3:6]}-{cleaned_number[6:]}" # Got external help (for me)
    else:
        return "Invalid Number"

raw_number = input().strip()
print(clean_number(raw_number)) 
```

### Question 11 · The Product SKU Inspector

An inventory system requires product codes (SKUs) to strictly match this format after being cleaned up: exactly **3 uppercase letters**, a single hyphen, and exactly **4 digits** (e.g. "ABC-1234").
Write a function that receives a messy SKU string. First clean it by stripping leading/trailing spaces and converting it to uppercase. Then validate it using string methods. Return "Valid: [Cleaned_SKU]" if it passes, or "Invalid SKU" if it fails any formatting rule.

> Note: Use isalpha(), isdigit() etc. — no re module. Case 2 fails (2 letters, 5 digits); Case 3 fails (special character).

```python
# Q11

def validate_sku(sku):
    sku = sku.strip().upper()

    if len(sku) != 8:
        return "Invalid SKU"

    if sku[3] != "-":
        return "Invalid SKU"

    if sku[:3].isalpha() and sku[4:].isdigit():
        return f"Valid: {sku}"

    return "Invalid SKU"


sku = input()
print(validate_sku(sku))
```
