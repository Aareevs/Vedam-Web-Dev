```python
secretcode = "charlie-bravo-alpha"
# Reverse order of words, convert every letter to uppercase and replace hyphe with spaces
lst = secretcode.split("-")
lst=lst[::-1]
ans = " ".join(lst)
ans = ans.upper()
print(ans)
```
